```python
import nltk
from nltk.stem import PorterStemmer

ps = PorterStemmer()

words = ["playing", "played", "plays", "studies",
         "studying", "connected", "connection",
         "happiness", "running", "quickly"]

print("PORTER STEMMING")
print("------------------------------")
print("Original Word\tStemmed Word")
print("------------------------------")

for word in words:
    stem = ps.stem(word)
    print(word, "\t\t", stem)
```
