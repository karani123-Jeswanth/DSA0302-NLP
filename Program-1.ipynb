```python
import re

def main():
    try:
        first_line = input()
        if not first_line.strip():
            return
        n = int(first_line.strip())
    except (EOFError, ValueError):
        return

    input_lines = []
    for _ in range(n):
        try:
            line = input()
        except EOFError:
            line = ""
        input_lines.append(line)

    pattern = re.compile(r"[A-Za-z]+(?:['-][A-Za-z]+)*")

    results = []
    for line in input_lines:
        tokens = pattern.findall(line)
        k = len(tokens)

        if k == 0:
            results.append("0")
        else:
            results.append(f"{k} " + " ".join(tokens))

    for res in results:
        print(res)

if __name__ == '__main__':
    main()
```
