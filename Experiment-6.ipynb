```python

# Program to generate plural nouns, verb forms,
# and comparative and superlative adjectives

# Irregular word lists
irregular_nouns = {
    "child": "children",
    "man": "men",
    "woman": "women",
    "foot": "feet",
    "tooth": "teeth",
    "mouse": "mice",
    "person": "people",
    "goose": "geese"
}

irregular_verbs = {
    "go": ("went", "going"),
    "eat": ("ate", "eating"),
    "write": ("wrote", "writing"),
    "run": ("ran", "running"),
    "swim": ("swam", "swimming"),
    "come": ("came", "coming"),
    "see": ("saw", "seeing"),
    "do": ("did", "doing")
}

irregular_adjectives = {
    "good": ("better", "best"),
    "bad": ("worse", "worst"),
    "far": ("farther", "farthest"),
    "little": ("less", "least")
}

# Plural form of nouns
def plural(noun):
    if noun in irregular_nouns:
        return irregular_nouns[noun]
    elif noun.endswith(("s", "sh", "ch", "x", "z")):
        return noun + "es"
    elif noun.endswith("y") and noun[-2] not in "aeiou":
        return noun[:-1] + "ies"
    elif noun.endswith("f"):
        return noun[:-1] + "ves"
    else:
        return noun + "s"

# Past tense and -ing form of verbs
def verb_forms(verb):
    if verb in irregular_verbs:
        return irregular_verbs[verb]

    # Past tense
    if verb.endswith("e"):
        past = verb + "d"
    elif verb.endswith("y") and verb[-2] not in "aeiou":
        past = verb[:-1] + "ied"
    elif (len(verb) > 2 and verb[-1] not in "aeiou"
          and verb[-2] in "aeiou"
          and verb[-3] not in "aeiou"):
        past = verb + verb[-1] + "ed"
    else:
        past = verb + "ed"

    # -ing form
    if verb.endswith("ie"):
        ing = verb[:-2] + "ying"
    elif verb.endswith("e") and not verb.endswith("ee"):
        ing = verb[:-1] + "ing"
    elif (len(verb) > 2 and verb[-1] not in "aeiou"
          and verb[-2] in "aeiou"
          and verb[-3] not in "aeiou"):
        ing = verb + verb[-1] + "ing"
    else:
        ing = verb + "ing"

    return past, ing

# Comparative and superlative forms of adjectives
def adjective_forms(adj):
    if adj in irregular_adjectives:
        return irregular_adjectives[adj]

    if adj.endswith("y"):
        comparative = adj[:-1] + "ier"
        superlative = adj[:-1] + "iest"
    elif adj.endswith("e"):
        comparative = adj + "r"
        superlative = adj + "st"
    elif len(adj) <= 5:
        comparative = adj + "er"
        superlative = adj + "est"
    else:
        comparative = "more " + adj
        superlative = "most " + adj

    return comparative, superlative

# Main program
print("WORD FORM GENERATOR")
print("-------------------")

noun = input("Enter a noun: ").lower()
verb = input("Enter a verb: ").lower()
adj = input("Enter an adjective: ").lower()

print("\nNOUN")
print("Plural form:", plural(noun))

past, ing = verb_forms(verb)
print("\nVERB")
print("Past tense:", past)
print("-ing form:", ing)

comparative, superlative = adjective_forms(adj)
print("\nADJECTIVE")
print("Comparative:", comparative)
print("Superlative:", superlative)
```
