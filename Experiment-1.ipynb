```python
import nltk
from nltk.util import ngrams
from nltk.tokenize import word_tokenize
from collections import Counter

nltk.download('punkt_tab')

sentence = input("Enter a sentence: ")

words = word_tokenize(sentence.lower())

print("\nUnigrams:")
print(list(ngrams(words, 1)))

print("\nBigrams:")
print(list(ngrams(words, 2)))

print("\nTrigrams:")
print(list(ngrams(words, 3)))

corpus = [
    "i like python",
    "i like machine learning",
    "i like data science",
    "i love python",
    "i love machine learning",
    "we like python",
    "we like data science"
]

bigram_counts = Counter()

for sentence in corpus:
    tokens = sentence.split()

    for i in range(len(tokens) - 1):
        bigram = (tokens[i], tokens[i + 1])
        bigram_counts[bigram] += 1

word = input("\nEnter a word to predict the next word: ").lower()

candidates = {}

for (first, second), count in bigram_counts.items():
    if first == word:
        candidates[second] = count

print("\nBigram Counts:")

if candidates:
    for next_word, count in candidates.items():
        print(word, "->", next_word, ":", count)

    predicted_word = max(candidates, key=candidates.get)
    print("\nPredicted next word:", predicted_word)
else:
    print("No prediction available for this word.")
```
