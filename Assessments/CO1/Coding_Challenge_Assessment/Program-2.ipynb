```python
def main():
    try:
        first_line = input()
        if not first_line.strip():
            return
        parts = first_line.split()
        n = int(parts[0])
        k = int(parts[1])

        alphabet = input().split()

        start_state = int(input().strip())

        acc_line = input().split()
        m = int(acc_line[0])
        accepting_states = set(int(x) for x in acc_line[1:])

        t = int(input().strip())
        transitions = {}
        for _ in range(t):
            trans_line = input().split()
            p = int(trans_line[0])
            c = trans_line[1]
            r = int(trans_line[2])

            key = (p, c)
            if key not in transitions:
                transitions[key] = set()
            transitions[key].add(r)

        q = int(input().strip())
        queries = []
        for _ in range(q):
            queries.append(input().strip())

    except (EOFError, IndexError, ValueError):
        return

    results = []
    for query in queries:
        if query == '#':
            current_states = {start_state}
        else:
            current_states = {start_state}
            for symbol in query:
                next_states = set()
                for state in current_states:
                    key = (state, symbol)
                    if key in transitions:
                        next_states.update(transitions[key])
                current_states = next_states
                if not current_states:
                    break

        if current_states.intersection(accepting_states):
            results.append("ACCEPT")
        else:
            results.append("REJECT")

    for res in results:
        print(res)

if __name__ == '__main__':
    main()
```
