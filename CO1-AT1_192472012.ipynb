```python
import re

resume = """
Name: Rahul Kumar
Email: rahul.kumar@gmail.com
Mobile: 9876543210
Skills: Python, Java, SQL, Machine Learning, NLP
Experience: 3 years
"""

# Extract name
name = re.search(r"Name:\s*(.*)", resume).group(1)

# Extract email
email = re.search(r"[\w.-]+@[\w.-]+\.\w+", resume).group()

# Extract mobile number
mobile = re.search(r"\b\d{10}\b", resume).group()

# Extract skills
skills_list = ["Python", "Java", "SQL", "Machine Learning", "NLP"]
skills = []

for skill in skills_list:
    if re.search(skill, resume, re.IGNORECASE):
        skills.append(skill)

# Extract experience
experience = re.search(r"Experience:\s*(\d+)\s*years?", resume, re.IGNORECASE)
years = int(experience.group(1))

# Display structured summary
print("----- Candidate Profile -----")
print("Name:", name)
print("Email:", email)
print("Mobile:", mobile)
print("Skills:", ", ".join(skills))
print("Experience:", years, "years")

# Eligibility
if years >= 2 and "Python" in skills:
    print("Eligibility: Eligible")
else:
    print("Eligibility: Not Eligible")
```

```python
import re

products = [
    "Apple iPhone 15",
    "Samsung Galaxy S24",
    "OnePlus Nord 2T",
    "Apple MacBook Air",
    "Dell Inspiron Laptop",
    "HP Pavilion Laptop",
    "Samsung Galaxy Tab",
    "OnePlus Buds"
]

keyword = input("Enter search keyword: ")

print("\nExact Match:")
count = 0
for product in products:
    if re.search(r"\b" + re.escape(keyword) + r"\b", product, re.IGNORECASE):
        print(product)
        count += 1
print("Total:", count)

print("\nPrefix Match:")
count = 0
for product in products:
    if re.search(r"\b" + re.escape(keyword), product, re.IGNORECASE):
        print(product)
        count += 1
print("Total:", count)

print("\nSuffix Match:")
count = 0
for product in products:
    if re.search(re.escape(keyword) + r"\b", product, re.IGNORECASE):
        print(product)
        count += 1
print("Total:", count)

print("\nPartial Match:")
count = 0
for product in products:
    if re.search(re.escape(keyword), product, re.IGNORECASE):
        print(product)
        count += 1
print("Total:", count)
```

```python
import re

register_no = input("Enter Register Number: ")
email = input("Enter Institutional Email: ")
course_code = input("Enter Course Code: ")
semester = input("Enter Semester: ")
mobile = input("Enter Mobile Number: ")

# Register number validation
if re.fullmatch(r"\d{2}[A-Z]{2}\d{4}", register_no):
    print("Register Number: Valid")
    reg_valid = True
else:
    print("Register Number: Invalid")
    reg_valid = False

# Email validation
if re.fullmatch(r"[a-zA-Z0-9._%+-]+@simats\.edu\.in", email):
    print("Institutional Email: Valid")
    email_valid = True
else:
    print("Institutional Email: Invalid")
    email_valid = False

# Course code validation
if re.fullmatch(r"[A-Z]{3}\d{4}", course_code):
    print("Course Code: Valid")
    course_valid = True
else:
    print("Course Code: Invalid")
    course_valid = False

# Semester validation
if re.fullmatch(r"[1-8]", semester):
    print("Semester: Valid")
    semester_valid = True
else:
    print("Semester: Invalid")
    semester_valid = False

# Mobile number validation
if re.fullmatch(r"[6-9]\d{9}", mobile):
    print("Mobile Number: Valid")
    mobile_valid = True
else:
    print("Mobile Number: Invalid")
    mobile_valid = False

# Final status
if reg_valid and email_valid and course_valid and semester_valid and mobile_valid:
    print("\nRegistration Status: Successful")
else:
    print("\nRegistration Status: Failed")
```
