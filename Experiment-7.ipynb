```python

# Program to identify derivational suffixes
# and show base word, suffix, POS change, and meaning

# Dictionary containing words and their derivational information
words = {
    "happiness": ("happy", "ness", "Adjective", "Noun",
                  "The state of being happy"),
    "kindness": ("kind", "ness", "Adjective", "Noun",
                 "The quality of being kind"),
    "teacher": ("teach", "er", "Verb", "Noun",
                "A person who teaches"),
    "quickly": ("quick", "ly", "Adjective", "Adverb",
                "In a quick manner"),
    "careful": ("care", "ful", "Noun", "Adjective",
                "Showing care or attention"),
    "careless": ("care", "less", "Noun", "Adjective",
                 "Without care or attention"),
    "development": ("develop", "ment", "Verb", "Noun",
                    "The process of developing"),
    "readable": ("read", "able", "Verb", "Adjective",
                 "Able to be read"),
    "national": ("nation", "al", "Noun", "Adjective",
                 "Related to a nation"),
    "friendship": ("friend", "ship", "Noun", "Noun",
                   "The relationship between friends")
}

word = input("Enter a word: ").lower()

if word in words:
    base, suffix, old_pos, new_pos, meaning = words[word]

    print("\nWORD ANALYSIS")
    print("----------------------")
    print("Original word:", word)
    print("Base word:", base)
    print("Derivational suffix:", suffix)
    print("Split form:", base + " + " + suffix)
    print("Base word POS:", old_pos)
    print("Derived word POS:", new_pos)
    print("Meaning:", meaning)

    if old_pos == new_pos:
        print("POS Change: No change in word class")
    else:
        print("POS Change:", old_pos, "to", new_pos)

else:
    print("Word not found in the dictionary.")
    print("Try: happiness, teacher, quickly, careful,")
    print("careless, development, readable, national,")
    print("kindness, or friendship.")
```
