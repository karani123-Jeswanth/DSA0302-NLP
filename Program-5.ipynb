```python
def is_consonant(word, i):
    ch = word[i]
    if ch in {'a', 'e', 'i', 'o', 'u'}:
        return False
    if ch == 'y':
        if i == 0:
            return True
        return not is_consonant(word, i - 1)
    return True

def get_word_form(word):
    form = []
    n = len(word)
    i = 0
    while i < n:
        if is_consonant(word, i):
            while i < n and is_consonant(word, i):
                i += 1
            form.append('C')
        else:
            while i < n and not is_consonant(word, i):
                i += 1
            form.append('V')
    return "".join(form)

def get_measure(stem):
    form = get_word_form(stem)
    return form.count('VC')

def has_vowel(stem):
    for i in range(len(stem)):
        if not is_consonant(stem, i):
            return True
    return False

def ends_double_consonant(stem):
    if len(stem) < 2:
        return False
    c1, c2 = stem[-1], stem[-2]
    return c1 == c2 and is_consonant(stem, len(stem) - 1) and is_consonant(stem, len(stem) - 2)

def ends_cvc(stem):
    if len(stem) < 3:
        return False
    c1 = stem[-3]
    v = stem[-2]
    c2 = stem[-1]
    return is_consonant(stem, len(stem)-3) and not is_consonant(stem, len(stem)-2) and is_consonant(stem, len(stem)-1) and c2 not in {'w', 'x', 'y'}

def porter_stemmer_step1(word):
    if word.endswith("sses"):
        word = word[:-2]
    elif word.endswith("ies"):
        word = word[:-2]
    elif word.endswith("ss"):
        pass
    elif word.endswith("s"):
        word = word[:-1]

    removed_suffix = False
    if word.endswith("eed"):
        stem = word[:-3]
        if get_measure(stem) > 0:
            word = stem + "ee"
    elif word.endswith("ed"):
        stem = word[:-2]
        if has_vowel(stem):
            word = stem
            removed_suffix = True
    elif word.endswith("ing"):
        stem = word[:-3]
        if has_vowel(stem):
            word = stem
            removed_suffix = True

    if removed_suffix:
        if word.endswith("at"):
            word = word + "e"
        elif word.endswith("bl"):
            word = word + "e"
        elif word.endswith("iz"):
            word = word + "e"
        elif ends_double_consonant(word) and word[-1] not in {'l', 's', 'z'}:
            word = word[:-1]
        elif get_measure(word) == 1 and ends_cvc(word):
            word = word + "e"

    if word.endswith("y") and len(word) > 1:
        stem = word[:-1]
        if has_vowel(stem):
            word = stem + "i"

    return word

def main():
    try:
        q_count = int(input().strip())
        words = []
        for _ in range(q_count):
            line = input().strip()
            if line:
                words.append(line)
    except (EOFError, ValueError):
        return

    results = [porter_stemmer_step1(w) for w in words]

    for res in results:
        print(res)

if __name__ == '__main__':
    main()
```
