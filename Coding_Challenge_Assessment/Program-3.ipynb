```python
def is_vowel(ch):
    return ch in {'a', 'e', 'i', 'o', 'u'}

def get_vowel_groups(word):
    groups = 0
    in_vowel_group = False
    for ch in word:
        if is_vowel(ch):
            if not in_vowel_group:
                groups += 1
                in_vowel_group = True
        else:
            in_vowel_group = False
    return groups

def is_doublable(stem):
    if len(stem) < 3:
        return False
    c1, v, c2 = stem[-3], stem[-2], stem[-1]
    if is_vowel(c1) or not is_vowel(v) or is_vowel(c2):
        return False
    if c2 in {'w', 'x', 'y'}:
        return False
    if get_vowel_groups(stem) != 1:
        return False
    return True

def apply_rules(stem, feature):
    if feature in {'N+PL', 'V+3SG'}:
        if stem.endswith(('s', 'x', 'z', 'ch', 'sh')):
            return stem + 'es'
        elif len(stem) >= 2 and not is_vowel(stem[-2]) and stem[-1] == 'o':
            return stem + 'es'
        elif len(stem) >= 2 and not is_vowel(stem[-2]) and stem[-1] == 'y':
            return stem[:-1] + 'ies'
        else:
            return stem + 's'

    elif feature == 'V+PROG':
        if len(stem) >= 3 and not is_vowel(stem[-1]) and stem[-1] != 'e' and stem[-2] in {'a', 'e', 'i', 'o', 'u'} and stem[-3] not in {'a', 'e', 'i', 'o', 'u'} and stem[-1] not in {'w', 'x', 'y'} and len(stem) >= 3: # Wait, let's use the explicit is_doublable check:
            pass

        if len(stem) >= 3 and not is_vowel(stem[-2]) and stem[-1] == 'e':
            return stem[:-1] + 'ing'
        elif is_doublable(stem):
            return stem + stem[-1] + 'ing'
        else:
            return stem + 'ing'

    elif feature in {'V+PAST', 'ADJ+COMP', 'ADJ+SUP'}:
        suffix_map = {
            'V+PAST': ('d', 'ed'),
            'ADJ+COMP': ('r', 'er'),
            'ADJ+SUP': ('st', 'est')
        }
        s1, s2 = suffix_map[feature]

        if stem.endswith('e'):
            return stem + s1
        elif len(stem) >= 2 and not is_vowel(stem[-2]) and stem[-1] == 'y':
            return stem[:-1] + 'i' + s2
        elif is_doublable(stem):
            return stem + stem[-1] + s2
        else:
            return stem + s2

    return stem

def main():
    try:
        l_count = int(input().strip())
        lexicon = {}
        for _ in range(l_count):
            line = input().split()
            if len(line) >= 3:
                stem, feature, form = line[0], line[1], line[2]
                lexicon[(stem, feature)] = form

        q_count = int(input().strip())
        queries = []
        for _ in range(q_count):
            line = input().split()
            if len(line) >= 2:
                queries.append((line[0], line[1]))

    except (EOFError, ValueError, IndexError):
        return

    results = []
    for stem, feature in queries:
        if (stem, feature) in lexicon:
            results.append(lexicon[(stem, feature)])
        else:
            results.append(apply_rules(stem, feature))

    for res in results:
        print(res)

if __name__ == '__main__':
    main()
```
