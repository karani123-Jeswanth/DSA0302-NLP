```python

# Morphological Analysis using NLTK

import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.tokenize import word_tokenize
from nltk import pos_tag
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet

# Input sentence
sentence = input("Enter a sentence: ")

# Tokenize the sentence
words = word_tokenize(sentence)

# Find POS tags
tags = pos_tag(words)

# Create lemmatizer
lemmatizer = WordNetLemmatizer()

print("\nMORPHOLOGICAL ANALYSIS")
print("-" * 85)
print(f"{'Word':<15}{'POS':<15}{'Lemma':<15}{'Number':<15}{'Tense':<15}")
print("-" * 85)

for word, tag in tags:

    # Convert NLTK POS tag to WordNet POS
    if tag.startswith('N'):
        pos = wordnet.NOUN
    elif tag.startswith('V'):
        pos = wordnet.VERB
    elif tag.startswith('J'):
        pos = wordnet.ADJ
    elif tag.startswith('R'):
        pos = wordnet.ADV
    else:
        pos = wordnet.NOUN

    # Find lemma
    lemma = lemmatizer.lemmatize(word, pos)

    # Find noun number
    if tag in ['NNS', 'NNPS']:
        number = "Plural"
    elif tag in ['NN', 'NNP']:
        number = "Singular"
    else:
        number = "-"

    # Find verb tense
    if tag == 'VBD' or tag == 'VBN':
        tense = "Past"
    elif tag in ['VBP', 'VBZ', 'VBG', 'VBP']:
        tense = "Present"
    elif tag == 'VB':
        tense = "Base"
    else:
        tense = "-"

    print(f"{word:<15}{tag:<15}{lemma:<15}{number:<15}{tense:<15}")
```
